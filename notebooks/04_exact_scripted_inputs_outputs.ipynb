The default selects the **revoked/no-prior-effect framework case at resume**. `MODEL_GROUP=0` is the supervisor in this archive; group 1 is the email subagent. The index preserves group and invocation position. It does **not** infer a global chronological sequence across groups.

The pause and resume files are separate phase captures. Repeated messages in a resumed input are the saved conversation context, not proof that those earlier calls ran again.

```python
CASE = "framework_prior0_revoked1"
PHASE = "resume"          # "pause" or "resume"
MODEL_GROUP = 0           # Select from the index; this snapshot uses supervisor=0, email=1.
INVOCATION = 0            # Each preserved group in this pilot has one recorded invocation per phase.
snapshot = traces.snapshot(CASE, PHASE)
pointer = f"/models/{MODEL_GROUP}/{INVOCATION}"
show(snapshot.view(pointer + "/role", "Exact recorded role"))
```

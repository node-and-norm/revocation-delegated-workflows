## 1. Two evidence trails, one careful account
```text
RESEARCH PROCESS
Assignment → produced artifact → critical review → human decision
   ↓                ↓                 ↓                 ↓
Exact text?     Version/hash?     Finding/correction?  Scope/approval?

EXPERIMENT INSIDE THAT PROCESS
Scripted supervisor → scripted email worker → local action record
           pause / checkpoint / resume → saved result
```
The arrows are a reading map, not an assertion that every historical handoff was logged. A saved output cannot fill a missing assignment. A decision summary cannot stand in for an exact transcript. A human’s approval to run a check is not validation of its conclusion.

## 5. What is missing, and how to improve the record
- **Not recovered:** complete original team assignment prompts, inherited context, all intermediate outputs and every reviewer exchange. No synthetic reconstruction is supplied.
- **Not consistently retained:** exact model/version for each research contribution, precise assignment timestamps and links binding each message to its artifact revision.
- **Not applicable:** hidden internal reasoning, inferred cognition, or invented personal names for generic experimental roles.
- **Private scope:** these records concern Mark’s private study. Publication requires separate privacy, rights, disclosure and claim review.

**Proposed future handoff record:** assignment text and time → permitted scope and supplied source versions → actual output/artifact hash → review findings → correction/version → decision and authorizing message. Record missing or interrupted outputs explicitly. Keep system instructions, credentials and unrelated private context out of any export. This is a proposed recording practice, not an implemented logging system.

The draft provides a foundation for a technical companion, not a claim that transparency gaps have all been closed.

### Proposed per-assignment ledger (not yet implemented)
| Field | Record when available | If unavailable |
|---|---|---|
| Assignment | Actual sent message, timestamp, permitted scope and role | Not recovered; do not reconstruct |
| Model | Provider/model/version actually reported for that contribution | Not recorded; never infer from role name |
| Input sources | Supplied file/version/hash and declared inherited context | Coverage unknown |
| Output | Actual returned message and artifact revision/hash, including rejection/error | Missing or interrupted; do not substitute a summary |
| Review | Actual reviewer assignment, findings, limitations and revised artifact | Summary only, or not recovered |
| Decision | Decision owner, selected action, evidence considered and permission scope | Pending or not recorded |
| Capture | Capture time, original event time if known, redactions and omissions | Unknown times stay unknown |

A completed row would cover one bounded assignment only. It would not prove complete history, independence, factual correctness or authority to take the next action. Record observable messages and artifacts, not private chain-of-thought.

# Did the withdrawal actually stop the action?

Node & Norm · Revocation in Delegated AI Workflows · Private practitioner guide · 24 November 2026

Someone notices a mistake in a payment they approved. They withdraw approval while the payment is still queued. The screen says the request was received. Did the payment stop?

The screen alone cannot answer. Policy tells a reviewer **who may intervene and when intervention should take effect**. Records from the system that performs the action tell the reviewer **whether that specific action went through**.

| What to check | Withdrawal took effect (S4) | Withdrawal did not reach the authority in time (S5) |
|:--|:--|:--|
| What did the requester know? | The request was accepted by the relay. No completion confirmation arrived by the end of observation. | Same. |
| What happened to permission? | The local authority committed the withdrawal. The relay did not forward the confirmation. | The relay held the withdrawal request. The previous permission remained active during the observed window. |
| What did the action system record? | It denied the later attempt as revoked. Zero simulated effects. | It recorded one authorized simulated effect while permission was still active. |

**Same requester status. Opposite action outcomes.** These were saved, scripted local cases. The fixture recorded an inert action. No real payment or email went out.

## 1. Reconstruct what happened

The **requester**, the **service that forwards the request**, and the **system that checks permission and records the outcome** may each see something different. Keep their records separate until you can link them.

- **S4:** Withdrawal recorded at event 7 → later action attempt received at event 8 → denied at event 9 → no simulated effect. The requester did not receive the completion message.
- **S5:** Relay held the withdrawal → action attempt received at event 6 → one simulated effect committed at event 7. The local authority still recognized the earlier permission. This was **not** a case of an action succeeding after the local withdrawal took effect.

Event numbers describe order **within** a case, not seconds, nor order across cases. In both the direct path and the LangGraph checkpoint-and-resume path, the same local authority and destination handled the action. Agreement across the two paths is not replication across independent enforcement systems.

## 2. Use the worksheet on one case

Choose one consequential action: a payment, notice, record update, or access change. Do not start with the assistant's message. Begin with the action the institution needs to account for.

| Write down | What to request |
|:--|:--|
| **Authority and effective trigger:** Who could withdraw approval, and when should it have taken effect? | Policy version, decision or delegation record, and the authority system's record for that scope. Distinguish when the withdrawal should have applied from when the system actually enforced it. |
| **Action and evidence owner:** What exact outcome are you trying to establish, and who holds that record? | The destination's authoritative transaction or audit record. For a payment, be explicit about the stage: queued, canceled, captured or settled. |
| **Match and window:** Do the records describe the same action and attempt? | Correlation IDs, scope, retry identities, and reliable ordering. Check the observation window, log coverage and retention before interpreting a missing entry. |
| **Finding and next step:** What can the current records justify? | Record receipt, effective withdrawal, destination outcome and requester confirmation separately. Cite the evidence. For anything unresolved, name the missing record, its likely owner and the next request. |

Use **blocked** for a matched denial or other affirmative prevention evidence with adequate coverage. Use **went through** for a matched effect. State whether it happened before or after withdrawal took effect only when the order is supported. Use **unresolved** when records cannot justify a conclusion. Report any later refund, correction or restoration separately. It does not turn an earlier effect into a prevented one.

**Practice:** Imagine the only record you have is a request acknowledgment. Write “outcome unresolved” and ask the payment operations team for the processor record linked to that payment ID and review window. A record for another transaction, an ambiguous status, or silence would not settle the outcome.

## 3. Finish the study in two stages

### First: finish the bounded study and the practitioner guide

Write one results account that links each claim to a saved record, reconciles the reports with the archived evidence, and distinguishes the request, the authority change, the effect and confirmation. Preserve the prior edition and its known limits. Keep LangGraph as the checkpoint-and-resume layer. The permission check belongs at the destination that commits the action. The case-report tool reads evidence; it does not enforce policy or prove the underlying records are authentic.

Before calling the guide ready, check Jupyter and browser layouts; check that the two cases, worksheet and exact source paths are easy to find. Ask a reviewer who did not build the study to follow the guide on a complete packet and an incomplete one, identify the outcome or leave it unresolved, and point to the record behind that conclusion. Treat an initial walkthrough as a usability check, not a measurement of reviewer benefit. Do not claim production reliability, native Temporal qualification, or successful reversal of an action from these cases.

### Then: evaluate TypeSafe Jev as a separate, optional evidence reviewer

Jev may help a reviewer locate evidence or identify what is missing in narrative records. It must not grant permission, control the destination, or certify whether an action stopped. Keep it out of the workflow. Exact joins and explicit event rules do not need a model; stop here if those rules solve the selected task.

- **Before any run:** Freeze a separate synthetic holdout, not the two teaching cases, with matched complete, withheld and contradictory records. Hide answer labels. Have an independently checked reference specify the supported outcome, exact record pointer or missing record, and when “unresolved” is required. Keep related packet variants together when splitting design and holdout cases.
- **Compare like with like:** Use the same packet, response contract and access budget for a strong source-linked checklist and Jev. Include a source-linked structured model baseline in any evaluation that aims to show a benefit specific to Jev. Give both a way to abstain, and preserve rejected or absent responses. Pin the model, SDK, prompts, packet versions, validation rules and denominators before examining holdout results.
- **Measure the costly errors:** Report false claims that an action was stopped, appropriate unresolved judgments, accurate evidence pointers or next-record requests, and invalid or missing output rates. Show counts over all scheduled packets, not only successful calls. A supported unresolved judgment is not the same as invalid output. A matched effect may establish that the action went through even when the total number of effects is unresolved. Validate a result against the supplied evidence, not against Jev's confidence alone.
- **Keep the trust boundary:** Send no private case files, secrets, or institutional data to an external provider. Do not use the separate Trust, Autonomy & Evidence Jev experiment as evidence of performance here. Any provider run needs its own agreed scope. A claim that Jev improves human review would still require a separate reviewer comparison.

**Proposed AI team split:** Evidence reviewer verifies source locators; methods reviewer challenges the baseline, holdout and scoring; implementer makes the smallest approved changes after the specification is fixed. Mark retains responsibility for conclusions and publication. This is a proposed evaluation, not a result.

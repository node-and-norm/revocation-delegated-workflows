## 2 · What was varied, and what was held fixed?

**Varied:** six explicit orderings of request delivery, fence installation, execution and confirmation, on direct and framework checkpoint/resume paths.

**Held fixed:** one trusted local sink, fixed root `rootR`, epoch 7 withdrawn to epoch 8, one parent-child grant, a valid ticket, no prior effect, one resumed execute attempt and at most one outstanding withdrawal per fresh case.

The direct and framework paths share the historical adapter/gateway. They test matched conformance, not two independently built security mechanisms. The gateway does not generically validate arbitrary revoke targets; the one-outstanding-request/fixed-authority fixture supplies the correlation boundary.

| Information boundary | Can observe | Cannot establish from that alone |
|---|---|---|
| Caller | Volatile acceptance, matching completion envelope, observation-end signal | Privileged sink state while unconfirmed. |
| Relay | Request and actual gateway response | Authenticated remote receipt or genuine network-loss behavior. |
| Evaluator/checker | Saved caller/relay traces, sink records, framework artifacts | Human cognition, hostile-worker confinement or population reliability. |

The caller does not receive the scenario label, expected answer, sink database or privileged barriers. This is a restricted interface within a trusted fixture, not an adversarial process sandbox.

## 7 · Reproduction readiness: what a future operator would need

1. **Scope and permission:** identify whether the task is artifact checking or new execution. This notebook authorizes neither new experiments nor previously denied Temporal operations.
2. **Complete archive:** verify bytes and preserve its layout, including `frozen_candidate_v02` and database companions. Use fresh output paths rather than modifying originals.
3. **Compatible environment:** recorded CPython 3.12 on Linux x86_64, exact package versions and hashed locks. The reproduction used an external pre-existing wheelhouse; the study ZIP does not contain every dependency.
4. **New output directories:** the documented runner refuses existing child output directories. Keep reproduction output separate from the original measured campaign.
5. **Evidence and checking plan:** retain commands, exit status, failure and cleanup information; recheck source hashes and compare deterministic fields. Generated process/checkpoint/message IDs may differ.
6. **Interpretation:** label the batch as reproduction, retain failures and do not expand the original campaign's denominator. No deployment reliability, response-loss or hostile-boundary claim follows.

For exact commands, use [the preserved reproduction instructions](../docs/review/WITHDRAWAL_INTERFACE_REPRODUCTION_v0.1.md). They are deliberately **not executable cells here**. No dependency installation, subprocess, socket, model/API call or workflow run is part of this notebook.

**Separate blocked extension:** native Temporal still needs its own permitted environment, exporter/capture, observation boundaries, outcome mapping and deadline/cleanup qualification. None of this chapter clears those gates.

## 2 · Follow the question

1. A caller requests withdrawal.
2. The trusted local sink may install a committed authority fence.
3. The caller may receive confirmation.
4. A worker may attempt a protected effect.

The study varies their order. A request receipt is not an installed fence, and an installed fence is not necessarily known to the caller.

[Inspect the six schedules and the executed scope](../docs/decisions/2026-11-03-request-to-fence-interface-study.md).

The caller sees the same uncertainty, but the sink records differ. An investigator needs the relevant evidence rather than a more confident summary.

## 6 · What complete means

For the **local twelve-case study**, completion means a frozen scope, preserved results and corrections, reproducible instructions, supported conclusions and explicit limits. Collection is already complete; the reader-facing account makes it inspectable.

For **Temporal**, completion requires the declared qualification gates and a separately approved measured protocol, or a documented decision to stop or defer. A blocked qualification is not a pass.

For **human evidence review**, completion requires its own feasible design, independent reference and approved evaluation, or a documented feasibility stop. The current packets do not establish reviewer benefit.

A research program can continue while a bounded study closes. Finishing does not require favorable results.

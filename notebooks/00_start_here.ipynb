## Choose a reading route

**New reader or organizational leader:** study report → 01's S4/S5 contrast → report limitations and completion gates. Start with the human distinction: requesting a stop, enforcing it and knowing it happened are different.

**Researcher or methods reviewer:** report → 03 → 01 → 02. Keep original observations, reproduction, qualification tests and AI-assisted reviews separate.

**Engineer or reproducibility reviewer:** 03 → 04 → source archives and implementation. Reading archived code does not execute it; reproduction is a separate authorized activity.

**Research-integrity or disclosure reviewer:** 02 → 04 → report's team/disclosure sections. Permission to proceed is not claim-level human validation, and retained publisher guidance is not a current policy check.

## Reading experience and evidence boundaries

The chapter navigation follows the clear book pattern of Node & Norm's [Catholic AI pilot](https://github.com/node-and-norm/control-evidence-corpus/tree/main/catholic-ai-pilot). That is a presentation reference, not a transfer of its findings or live service observations into this study.

Here, the v0.3 messages are deterministic scripted outputs from actual framework machinery. The complete static invocation view in 04 requires no live controls. Other notebooks retain saved representative views and optional selectors; they do not claim to display every underlying archive record inline.

Dates, hashes and source locators preserve version boundaries. Missing material remains missing. AI helped prepare and review the chapters; separate AI-assisted checks are not independent human validation. The repository remains private and no public release is implied.

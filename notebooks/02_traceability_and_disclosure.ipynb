### Technical and AI-use disclosure

A small standard-library OOXML reader uses the workbook's own table definitions, not guessed row limits. It rejects a changed workbook digest and unsupported formulas. It never follows external relationships. Blank cells remain explicit. Numeric Excel dates are decoded for display; raw cell values remain in `book.cells` for inspection. Cells outside the named tables, including the future template and section headings, are not turned into records. This is not a general Excel rendering engine.

AI assisted implementation, documentation and review. The original workbook, its wording, source dates and private identifiers are preserved. This remains a private research artifact; public sharing needs a separate rights, claims and identifier review.
